```python
import pandas as pd

# Load the cleaned box office data from last week
mfile_path = r'C:/Users/dangb/Downloads/enhanced_box_office_data(2000-2024)u.csv'
movies = pd.read_csv(mfile_path)

# Load the Marvel Movies dataset
mmfile_path = r'C:/Users/dangb/Downloads/archive/Marvel Movies.csv'
marvelmovies = pd.read_csv(mmfile_path)
```

```python
# Merge the datasets on 'Year' (box office) and 'year' (Marvel movies)
merged_data = movies.merge(marvelmovies, left_on='Year', right_on='year', how='left')
```

```python
# Save the merged dataset
merged_data.to_csv("merged_box_office_marvel.csv", index=False)
```

### Explanation
I used a left join to merge the datasets. This ensures that all rows from the `enhanced_box_office` dataset are retained, and matching rows from the `Marvel Movies` dataset are added where available.

If there are years in the box office dataset that do not have corresponding records in the Marvel dataset, those rows will have `NaN` values in the Marvel-specific columns. This prevents loss of any box office data while still incorporating available Marvel movie details.
